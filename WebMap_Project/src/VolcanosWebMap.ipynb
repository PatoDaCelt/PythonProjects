```python
import folium
import pandas as pd
```

```python
data = pd.read_csv("Volcanoes.txt")
data
```

```python
lat = list(data["LAT"])
lon = list(data["LON"])
elev = list(data["ELEV"])
name = list(data["NAME"])

html = """<h4>Volcano information:</h4>
Name: %s
<br>
Height: %s m
"""
```

```python
def choose_color(elev):
    if elev < 500:
        color = "darkgreen"
    elif elev < 1000:
        color = "green"
    elif elev < 1500:
        color = "lightgreen"
    elif elev < 2000:
        color = "yellow"
    elif elev < 2500:
        color = "orange"
    elif elev < 3000:
        color = "red"
    elif elev < 3500:
        color = "darkred"
    elif elev < 4000:
        color = "purple"
    else:
        color = "black"
        
    return color
```

```python
map = folium.Map(
    location=[38.58, -99.09],
    zoom_start=6,
    # tiles="Mapbox Bright"
)

#Grupo
fg = folium.FeatureGroup(name="My Map")

for lt, ln, el, nm in zip(lat, lon, elev, name):
    iframe = folium.IFrame(html=html % (nm, el), width=200, height=100)
    
    fg.add_child(folium.CircleMarker(
        location=[lt,ln],
        popup= folium.Popup(iframe),
        fill_color=choose_color(el),
        color="grey",
        fill_opacity=1,
        radius=5
    ))
```

```python
import json
with open('world.json', 'r', encoding='utf-8-sig') as f:
    world_data = json.load(f)

fg.add_child(folium.GeoJson(
    data=world_data,
    style_function=lambda x: {'fillColor': 'yellow'}
))
```

```python
map.add_child(fg)
map.save("Map1.html")
```
