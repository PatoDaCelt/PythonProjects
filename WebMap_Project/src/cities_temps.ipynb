```python
import folium
import pandas as pd
```

```python
data = pd.read_csv("cities.csv")
data
```

```python
lat = list(data["Lat"])
lon = list(data["Lon"])
name = list(data["City"])
temp = list(data["Temp"])

html = """
Ciudad: %s
<br>
Temperatura: %s°C
"""
```

```python
def color_temp(temp):
    if temp < 10:
        return "blue"
    elif temp < 15:
        return "lightblue"
    elif temp < 20:
        return "green"
    elif temp < 25:
        return "yellow"
    elif temp < 30:
        return "orange"
    elif temp < 40:
        return "red"
    else:
        return "black"
```

```python
map = folium.Map(
    location=[20, 0],
    zoom_start=2,
    # tiles="Mapbox Bright"
)
```

```python
#Grupo
fgt = folium.FeatureGroup(name="Temperaturas")

for lt, ln, nm, tp in zip(lat,lon,name,temp):
    iframe = folium.IFrame(html=html % (nm,tp), width=200, height=100)
    
    fgt.add_child(folium.CircleMarker(
        location=[lt,ln],
        popup = folium.Popup(iframe),
        fill_color = color_temp(tp),
        color ="grey",
        fill_opacity = 1,
        radius = 5
    ))
```

```python
map.add_child(fgt)
map.add_child(folium.LayerControl())

map.save("CitiesTempsMap.html")
```
