```python
import pandas as pd
import numpy as np
```

    Selecting

Se refiere a seleccionar columnas o filas en un DataFrame usando índices, etiqeutas o condiciones.

```python
df = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie", "David"],
    "Age": [25, 30, 35, 40],
    "City": ["NY", "Paris", "London", "Berlin"]
})

#Seleccionar filas por índice (iloc -> por posición)
print(df.iloc[0:2])
print()

#Seleccionar filas por etiqueta (loc -> por nombre de índice)
print(df.loc[0:2, ["Name"]])
```

```python
df1 = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie", "David", "Emilio"],
    "Age":  [25, None, 35, 40, 28],
    "City": ["NY", "Paris", None, "Berlin", "London"]
})

#.isnull() Devuelve True si el valor es nulo (NaN o None).
print(df1[df1["Age"].isnull()]) #Filas donde Age está vacío
print()

#.notnull() Devuelve True si el valor no es nulo.
print(df1[df1["City"].notnull()]) #Devuelve True si el valor no es nulo
print()

#.any() Verifica si al menos un valor es True en un eje.
print(df1.isnull().any()) #Ver si hay algún nulo por columna.
print()

#.all() Verifica si todos los valores son True en un eje.
print(df1.isnull().all()) #Verifica que columnas tienen todos los valores nulos.
```

    Indexing with isin

Se usa para filtrar filas basandose en si los valores de una columna están dentro de una lista de opciones.

```python
#Selecionar filas donde City es "London" o "Paris"
filtered = df[df["City"].isin(["London", "Paris"])]
filtered
```

    Where

Permite aplicar una condición, por ejemplo:

Los valores que cumplen se mantienen, los que no se reemplazan por NaN (a menos que especifiques otro valor con other=).

```python
print(df.where(df["Age"] > 30, other=0))
```

    Query

Permite filtrar datos escribiendo condiciones como una expresión de SQL dentro de un string.
Es más legible y conviene para consultas complejas.

```python
#Filtrar filas donde Age > 25 y City = 'Paris'
print(df.query("Age > 25 and City == 'Paris'"))
```

    Setting and Resetting Index

El índice en pandas no siempre tiene que ser 0,1,2,etc.

Puedes configurar cualquier columna como índice y luego resetearlo cuando lo necesites.

```python
#Establecer "Name" como índice
dfx = df.set_index("Name")
print(dfx)

print()

#Resetear el índice (volver al indice numerico por defecto)
dfg = dfx.reset_index()
print(dfg)
```

    Reindexing 

reindex() sirve para cambiar el indice de un DataFrame o Serie

-Puedes agregar/quitar indices

-Los valores que no existan se llenan con NaN.

```python
print("Original:\n", df)

# Reindex con índices nuevos
new_index = [1, 2, 3, 4, 5]
df2 = df.reindex(new_index)
print("\nReindexed:\n", df2)
```

    Forward Filling
ffill() Llena valores faltantes (NaN) con el último valor válido conocido.

```python
df2_ffill = df2.ffill()
df2_ffill
```

    Backward Filling
bfill() Llena valores faltantes con el proximo valor válido conocido.

```python
df2_bfill = df2.bfill()
df2_bfill
```

    MultiIndexing
Permite tener multiples niveles de índices en filas o columnas.

```python
arrays = [
    ['USA', 'USA', 'Canada', 'Canada'],
    ['NY', 'California', 'Toronto', 'Montereal']
]

#Ejemplo MultiIndex en filas
index = pd.MultiIndex.from_arrays(arrays, names=("Country", "City"))
df3 = pd.DataFrame({"Population": [0.5, 39.5, 2.9, 1.7]}, index=index)
print(df3)
```

```python
df4 = pd.DataFrame({
        ("Sales", "2022"): [100, 200],
        ("Sales", "2023"): [150, 250],
        ("Profit", "2022"): [30, 50],
        ("Profit", "2023"): [40, 60],
    },
    index=["Product A", "Product B"]
)

print(df4)
```
