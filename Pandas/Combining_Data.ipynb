```python
import pandas as pd
```

    Duplcate Data   
Los datos duplicados son filas o valores repetidos en un DataFrame.

.duplicate() -> marca si la fila ya apareció antes.

.drop_duplicates() -> elimina duplicados.

```python
df = pd.DataFrame({
    "Name": ["Alice", "Bob", "Alice", "David"],
    "Age": [25, 30, 25, 40]
})

print(df.duplicated())
print()
print(df.drop_duplicates())
```

    Grouping Data
grouphy() -> Agrupa datos por categoria y aplica funciones de agregación (sum, mean, count, etc.)

```python
df1 = pd.DataFrame({
    "City": ["NY", "NY", "Paris", "Paris", "London"],
    "Sales": [100, 200, 150, 250, 100]
})

grouped = df1.groupby("City")["Sales"].sum()
print(grouped)
```

    Missing Data
Valores faltantes (NaN). Pandas ofrece métodos para detectarlos y manejarlos.

.isnull() -> detecta nulos.
.notnull() -> detecta no nulos.
.fillna(valor) -> reemplaza nulos por un valor que le des.
.dropna() -> elimina filas/columnas con nulos.

```python
df2 = pd.DataFrame({
    "Name": ["Alice", "Bob", None],
    "Age": [25, None, 30]
})

print(df2.isnull())      #Ver que es nulo
print()
print(df2.fillna("Unknown")) # Rellenar
print()
print(df2.dropna())     # Eliminar
```

    Combining Data
Combinar DataFrames diferentes como Merge, Join, Concat.


        Merge
Muy parecido a SQL JOIN, combina dos DataFrames por columnas comunes.

how="inner" (por defecto) -> intersección

how="outer" -> unión completa

how="left" o right

```python
df3 = pd.DataFrame({
        "ID": [1,2,3,4],
        "Name": ["Alice", "Bob", "Charlie", "David"]
})
df4 = pd.DataFrame({
        "ID": [1,2,4,5],
        "Name": ["NY", "Paris", "London", "Sydney"]
})

merged = pd.merge(df3, df4, on="ID", how="inner")
print(merged)
```

        Join
Es un atajo de merge, usado cuando los índices son claves de unión.

```python
df5 = pd.DataFrame({
        "Name": ["Alice", "Bob", "Charlie", "David"]}, index=[1,2,3,4])
df6 = pd.DataFrame({
        "City": ["NY", "Paris", "London", "Sydney"]}, index=[1,2,4,5])

joined = df5.join(df6)
print(joined)
```

        Concatenate
Une DataFrames uno debajo de otro (filas) o uno al lado del otro (columnas).

```python
df7 = pd.DataFrame({"A": [1, 2], 
                    "B": [3, 4]})
df8 = pd.DataFrame({"A": [5, 6], 
                    "B": [7, 8]})

#Conactenar por Filas
concat_row = pd.concat([df7, df8])
print(concat_row)

print()

#Concatenar por Columna
concat_col = pd.concat([df7, df8], axis=1)
print(concat_col)
```
