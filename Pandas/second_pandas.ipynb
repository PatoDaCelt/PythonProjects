El método pivot() se usa para reorganizar un DataFrame cambiando filas a columnas.

Usa valores únicos de una columna y los convierte en encabezados.

```python
df = pd.DataFrame({
    "date": ["2023-01", "2023-01", "2023-02", "2023-02",],
    "city": ["London", "Paris", "London", "Paris"],
    "sales": [100, 120, 90, 110]
})

pivoted = df.pivot(index="date", columns="city", values="sales")

#df
pivoted
```

El método pivot_table() es parecido a pivot pero más flexible, permite adiciones aun habiendo valores repetidos.

Usa funciones como mean(), sum(), count(), etc.

```python
df1 = pd.DataFrame({
    "city": ["London", "Paris", "London", "Paris"],
    "month": ["Jan", "Jan", "Feb", "March"],
    "sales": [100, 120, 90, 110]
})

pt = pd.pivot_table(df1, index="city", columns="month", values="sales", aggfunc="mean")

df1
```

stack() -> mueve columnas al índice (filas)

unstack() -> mueve índices al nivel de columnas

```python
data = {
    "A": [1, 2],
    "B": [3, 4]
}

df3 = pd.DataFrame(data, index=["rox1", "rox2"])

df3
```

```python
stack_table = df3.stack()
stack_table
```

```python
unstack_table = stack_table.unstack()
unstack_table
```

Iterar en Pandas significa recorrer filas o columnas de un DataFrame.
los más comúnes son:

    -iterrows() -> itera fila por fila
    -itertuples() -> itera fila por fila, pero como tuplas
    -items() -> itera columna por columna

```python
df4 = pd.DataFrame({
    "Name": ["Alice", "Bob", "Pato", "Leslie", "Maria", "Juan"],
    "Age": [25, 30, 22, 16, 45, 52]
})

#Iterar filas
for i, row in df4.iterrows():
    print(i, row["Name"], row["Age"])
    
print() #Espacio blanco

#Iterar columnas
for columns, values in df4.items():
    print(columns, list(values))
```

El método melt() agrupa las columnas en filas.

```python
df5 = pd.DataFrame({
    "Date": ["2016-03-01", "2016-03-02", "2016-03-01", "2016-03-03", "2016-03-02", "2016-03-03"],
    "Type": ["a", "b", "c", "a", "a", "c"],
    "Values": [11.43, 13.03, 20.78, 99.90, 1.34, 20.78]
})

melt_table = pd.melt(df5, id_vars="Date", value_vars=["Type", "Values"], value_name="Observations")

melt_table
```
